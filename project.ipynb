```python
import networkx as nx
import matplotlib.pyplot as plt

G = nx.DiGraph()

# Add nodes
G.add_nodes_from(["joy", "sadness", "anger", "fear", "love", "surprise"])

# Relations from LLM
relations = [
    ("sadness", "IS_OPPOSITE_OF", "joy"),
    ("anger", "IS_OPPOSITE_OF", "love"),
    ("joy", "IS_A_TYPE_OF", "love"),
    ("fear", "CAN_COEXIST_WITH", "sadness"),
    ("anger", "CAUSES", "sadness"),
    ("fear", "CAN_TRIGGER", "anger"),
    ("joy", "CAUSES", "surprise"),
    ("surprise", "IS_A_CATALYST_FOR", "joy"),
    ("surprise", "IS_A_CATALYST_FOR", "fear"),
    ("love", "MITIGATES", "fear")
]

# Add relationships
for node1, relation, node2 in relations: 
    G.add_edge(node1, node2, relation=relation)


# Visualize
pos = nx.spring_layout(G)
nx.draw(G, pos, with_labels=True, node_color="lightblue", node_size=3000, font_size=10)
edge_labels = nx.get_edge_attributes(G, 'relation')
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels)
plt.show()
```

```python
# print(G.nodes())
for rel in G.edges(data=True):
    print(rel)
```
